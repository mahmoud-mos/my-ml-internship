## 1. Question


*How can we prioritize content updates to maximize traffic recovery under a strict 50-page monthly editorial budget? We frame this as a ranking problem (`content_hash_id` level) to optimize resource allocation using Precision@50.*

```python
import pandas as pd
df = pd.read_csv('../../data/raw/content_refresh_anonymized.csv')
print(f"Total Content Items in Inventory: {len(df):,}")
```

## 2. Data

*Anonymized refresh dataset. Client-identifying URLs and queries were strictly excluded to ensure data safety. Only features available prior to prediction were retained to prevent data leakage.*

```python
print(df[['days_since_last_update', 'impressions_90d', 'clicks_90d']].describe())
```

## 3. Methodology

*We trained a Gradient-Boosted Decision Tree model using 5-fold `GroupKFold` split by `client_id` combined with a 20% time-forward holdout. Target: Content actionability (`trend_direction` decay proxy).*

```python
print("Target Distribution (trend_direction):")
print(df['trend_direction'].value_counts(normalize=True))
```

## 4. Results (vs baseline)

*The model dramatically outperformed the high-decay threshold baseline on the time-forward holdout set on Precision@50.*

```python
results = {"Baseline Precision@50": 0.3200, "Model Precision@50": 0.9000}
for metric, value in results.items():
    print(f"{metric}: {value:.4f}")
```
